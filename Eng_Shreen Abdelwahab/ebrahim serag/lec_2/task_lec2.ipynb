```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
x = np.array([2, 4, 6, 8, 10])
y = np.array([40, 50, 65, 80, 90])
n = len(x)
x_mean = x.sum() / n
y_mean = y.sum() / n
print(f"x_mean = {x_mean}, y_mean = {y_mean}")
```

```python
sxy = ((x - x_mean) * (y - y_mean)).sum()
sxx = ((x - x_mean) ** 2).sum()
b1 = sxy / sxx
b0 = y_mean - b1 * x_mean
print(f"Slope b1 = {sxy:.0f}/{sxx:.0f} = {b1}")
print(f"Intercept b0 = {y_mean} - {b1}*{x_mean} = {b0}")
print(f"y_hat = {b0} + {b1}x")
```

```python
y_hat = b0 + b1 * x
e = y - y_hat
print(" x   y   y_hat   e")
for xi, yi, yh, ei in zip(x, y, y_hat, e):
    print(f"{xi:2d} {yi:3d} {yh:6.1f} {ei:+5.1f}")
print(f"residuals = {e.sum():.1f}")
```

```python
mae  = np.abs(e).sum() / n
mse  = (e ** 2).sum() / n
rmse = np.sqrt(mse)
sse  = (e ** 2).sum()
sst  = ((y - y_mean) ** 2).sum()
r2   = 1 - sse / sst
print(f"MAE = {mae:.2f} | MSE = {mse:.2f} | RMSE = {rmse:.2f} | R2 = {r2:.4f}")
print(f"studen study 7 hour {b0 + b1 * 7}")
```

```python
model = LinearRegression().fit(x.reshape(-1, 1), y)
pred = model.predict(x.reshape(-1, 1))
print("scikit-learn")
print("intercept =", round(model.intercept_, 1), "slope =", round(model.coef_[0], 1))
print("MAE =", round(mean_absolute_error(y, pred), 2))
print("MSE =", round(mean_squared_error(y, pred), 2))
print("R2 =", round(r2_score(y, pred), 4))
print("predict(7) =", round(model.predict([[7]])[0], 1))
print(20*"*")
print("MSE Comparison")
for name, c0, c1 in [("Best fit", 26, 6.5), ("Small slope", 26, 3), ("Large intercept", 50, 6.5)]:
    mse_try = ((y - (c0 + c1 * x)) ** 2).mean()
    print(name, "b0 =", c0, "b1 =", c1, "-> MSE =", round(mse_try, 1))
```
