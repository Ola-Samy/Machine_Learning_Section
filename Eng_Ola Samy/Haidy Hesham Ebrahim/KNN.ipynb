```python


# 1. Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

# 2. Create Dataset
X = np.array([
    [1, 2],
    [2, 1],
    [2, 3],
    [3, 2],
    [6, 5],
    [7, 7],
    [8, 8],
    [7, 5]
])

y = np.array([
    "Fail", "Fail", "Fail", "Fail",
    "Pass", "Pass", "Pass", "Pass"
])

# 3. New Student
new_student = np.array([5, 4])

# 4. Choose K
K = 3

# 5. Calculate Euclidean Distance
distances = []

for i in range(len(X)):
    distance = np.sqrt(np.sum((X[i] - new_student) ** 2))
    distances.append((distance, y[i], i))

# 6. Sort Distances
distances.sort(key=lambda item: item[0])

# 7. Select K Nearest Neighbors
nearest_neighbors = distances[:K]

# 8. Display Nearest Neighbors
print("Nearest Neighbors:")

for distance, label, index in nearest_neighbors:
    print(f"P{index + 1}: Distance = {distance:.2f}, Class = {label}")

# 9. Majority Voting
votes = Counter(label for distance, label, index in nearest_neighbors)

prediction = votes.most_common(1)[0][0]

print("\nFinal Prediction:", prediction)

# 10. Draw the Data
plt.scatter(X[y == "Pass", 0], X[y == "Pass", 1],
            color="green", label="Pass", s=100)

plt.scatter(X[y == "Fail", 0], X[y == "Fail", 1],
            color="red", label="Fail", s=100)

plt.scatter(new_student[0], new_student[1],
            color="blue", marker="*", s=250, label="New Student")

# Draw lines to nearest neighbors
for distance, label, index in nearest_neighbors:
    plt.plot([5, X[index, 0]], [4, X[index, 1]], "k--")

plt.xlabel("Study Hours")
plt.ylabel("Attendance")
plt.title("KNN Classification")
plt.legend()
plt.grid()
plt.show()
```
