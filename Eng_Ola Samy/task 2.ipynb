```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
data = {
    "annual_income": [
        15, 18, 20, 22, 25,
        40, 42, 45, 48, 50,
        70, 75, 78, 82, 85
    ],

    "spending_score": [
        80, 75, 85, 78, 82,
        40, 45, 38, 42, 48,
        20, 25, 18, 22, 28
    ]
}
df = pd.DataFrame(data)
df
```

```python

scaler= StandardScaler()
x = scaler.fit_transform(df)
```

```python
model = KMeans(n_clusters=2, random_state=0, n_init=10 )
df['cluster'] = model.fit_predict(x)
df
```

```python
print('customer grouping center', model.cluster_centers_)
```

```python
print('Silhouette Score',silhouette_score(x,df['cluster']))
```

```python
plt.scatter(df['annual_income'], df['spending_score'], c=df['cluster'])
plt.xlabel('annual_income')
plt.ylabel('spending_score')
plt.title('customer segmentaion')
plt.tight_layout()
plt.show()
```
